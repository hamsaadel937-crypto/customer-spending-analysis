```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('customer_spending.csv')
df.head()
```

## 1. Explore the dataset
We inspect the shape, data types, and missing values before analysis.

```python
print('Shape:', df.shape)
print(df.dtypes)
print('\nMissing values:')
print(df.isnull().sum())
```

## 2. Clean missing values
Median is used for numeric variables because it is less affected by extreme values. Mode is used for the categorical city column.

```python
df['Annual_Income'] = df['Annual_Income'].fillna(df['Annual_Income'].median())
df['Satisfaction_Score'] = df['Satisfaction_Score'].fillna(df['Satisfaction_Score'].median())
df['City'] = df['City'].fillna(df['City'].mode()[0])
print(df.isnull().sum())
```

## 3. Descriptive statistics

```python
df.describe()
```

## 4. Feature engineering
Average Order Value helps us understand the average spending associated with each order.

```python
df['Average_Order_Value'] = df['Annual_Spending'] / df['Orders'].replace(0, 1)
df[['Orders','Annual_Spending','Average_Order_Value']].head()
```

## 5. Spending by city

```python
city_spending = df.groupby('City')['Annual_Spending'].mean().sort_values(ascending=False)
print(city_spending)
city_spending.plot(kind='bar', figsize=(8,5), title='Average Spending by City')
plt.ylabel('Average Annual Spending')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 6. Correlation analysis
Correlation helps us measure the strength and direction of linear relationships between numerical variables. Correlation does not prove causation.

```python
cols = ['Annual_Income','Website_Visits','Orders','Satisfaction_Score','Annual_Spending']
corr = df[cols].corr()
print(corr['Annual_Spending'].sort_values(ascending=False))
```

## 7. Income vs spending

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Annual_Income'], df['Annual_Spending'], alpha=0.7)
plt.xlabel('Annual Income')
plt.ylabel('Annual Spending')
plt.title('Income vs Annual Spending')
plt.tight_layout()
plt.show()
```

## 8. Key insights
The analysis can be summarized using the city average and the correlations with annual spending.

```python
top_city = city_spending.index[0]
print(f'Highest-spending city: {top_city}')
print(f'Income correlation: {corr.loc["Annual_Income", "Annual_Spending"]:.2f}')
print(f'Orders correlation: {corr.loc["Orders", "Annual_Spending"]:.2f}')
```
